```python
import sys
from pathlib import Path

if Path.cwd().name == "notebooks":
    import os
    os.chdir("..")
sys.path.insert(0, "src")

from data import load_config, get_processed_corpus
from models import BM25Retriever, TFIDFRetriever
```

```python
cfg = load_config()
corpus = get_processed_corpus(cfg, variant="v1")
```

```python
# == BASELINE ==
bm25_retriever = BM25Retriever(corpus)
tfidf_retriever = TFIDFRetriever(corpus)

# Test query
query = "What is federalism and how are powers divided between states and government?"

# BM25
bm25_results = bm25_retriever.search(query, top_k=5)
print("=== BM25 Results ===")
for i, res in enumerate(bm25_results, 1):
    print(f"{i}. [Ch: {res['chapter']}, Sec: {res['section']}, Page: {res['printed_page']}]")
    print(res["text"][:200] + "...\n")

# TF-IDF
tfidf_results = tfidf_retriever.search(query, top_k=5)
print("=== TF-IDF Results ===")
for i, res in enumerate(tfidf_results, 1):
    print(f"{i}. [Ch: {res['chapter']}, Sec: {res['section']}, Page: {res['printed_page']}] (Score: {res['score']:.4f})")
    print(res["text"][:200] + "...\n")
```
